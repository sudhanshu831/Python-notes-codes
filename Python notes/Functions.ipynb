```python
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    print(result)
factorial(5)
```

```python
def count_vowels():
    text = input("enter a word")
    vowels = "aeiouAEIOU"
    count = 0
    for char in text:
        if char in vowels:
            count += 1
    return count
    
count_vowels()
```

```python
def palindrome():
    p = input("enter a word")
    if (p == p[::-1]):
        print("it is a pallindrome")
    else:
        print("it is not a palindrome")
palindrome()
```

```python
def find_max():
    nums = input("Enter numbers separated by spaces: ")

    l = [int(x) for x in nums.split()]

    print("Maximum number is:", max(l))

find_max()
```

```python
def find_min():
    nums = input("Enter numbers separated by spaces: ")
    l = [int(x) for x in nums.split()]
    print("Minimum number is:", min(l))

find_min()
```

```python
def sum_list():
    numbers = input("Enter numbers separated by spaces: ")
    
    l = [int(x) for x in numbers.split()]
    
    total = 0
    
    for i in l:
        total += i
    
    print("The sum is:", total)

sum_list()
  
```

```python
def average():
    num = input("enter numbers separated by spaces:")
    l = [int (x) for x in num.split()]

    avg = 0

    for i in l:
        avg = (avg + i)
    print("the average is:", avg/len(l))

average()
```
