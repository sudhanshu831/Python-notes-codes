```python
import random

def play_game():
    print("=" * 40)
    print(" Welcome to the Number Guessing Game! ")
    print("=" * 40)
    
    # 1. Set the range and pick a random target
    low, high = 1, 100
    secret_number = random.randint(low, high)
    
    # 2. Choose difficulty level
    print("\nSelect a difficulty:")
    print("1. Easy (10 attempts)")
    print("2. Medium (7 attempts)")
    print("3. Hard (4 attempts)")
    
    attempts_map = {"1": 10, "2": 7, "3": 4}
    choice = input("Enter 1, 2, or 3: ").strip()
    max_attempts = attempts_map.get(choice, 7)
    
    print(f"\nI'm thinking of a number between {low} and {high}.")
    print(f"You have {max_attempts} attempts. Good luck!\n")
    
    # 3. Main game loop
    attempts_taken = 0
    while attempts_taken < max_attempts:
        raw_guess = input(f"[Attempt {attempts_taken + 1}/{max_attempts}] Enter your guess: ")
        
        # Validate that the input is an integer
        try:
            guess = int(raw_guess)
        except ValueError:
            print("Please enter a valid whole number.")
            continue
            
        attempts_taken += 1
        
        # Compare guess against secret number
        if guess < secret_number:
            print("Too low!")
        elif guess > secret_number:
            print("Too high!")
        else:
            print(f"\nBullseye! You found the number in {attempts_taken} attempts.")
            return
            
    print(f"\nGame over! The secret number was {secret_number}.")

if __name__ == "__main__":
    while True:
        play_game()
        play_again = input("\nWould you like to play again? (y/n): ").strip().lower()
        if play_again != "y":
            print("Thanks for playing! Goodbye.")
            break
```
