***File I/O***

```python
f = open("a.txt", "r")        # Opens a file and reads only
```

```python
line1 = f.readline()           # this reads only one line
print(line1)

data = f.read()                # this reads the whole doc
print(data)
f.close
```

```python
f = open("a.txt", "w")            #opens the file and writes in it
f.write("I want to be successful")
f.close
```

```python
f = open("b.txt", "a")            # opens the file and appends in it
f.write("I want to be successful")   # if a file doesn't exit it creates a new file
f.close
```

```python
f = open("b.txt", "r+")            # reads and overwrites
data = f.read()
print(data)
f.write("I want to get a job")
f.close
```

```python
with open ("a.txt", "r") as f:
    data = f.read()
    print(data)
```

```python
with open ("a.txt","w") as f:
        f.write("new data")
```
