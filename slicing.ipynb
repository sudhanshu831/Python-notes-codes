***Indexing***

```python
a = [1,2,3,9,7,6,5,4,0]
print(a[6])
```

```python
print(a[-1])
```

***Slicing***

```python
print(a[3:6:1])
```

```python
print(a[2:8])
print(a[2:8:2])
print(a[:8])
print(a[5:])
print(a[::2])
print(a[::-1])
```

```python
a = [1,2,3,9,7,6,5,4,0]
print(a[::-2])
```
