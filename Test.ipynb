```python
products = []

while True:
    print("\nMenu:")
    print("1. Add Product")
    print("2. Remove Product")
    print("3. Search Product")
    print("4. Display Products")
    print("5. Exit")

    try:
        choice = int(input("Enter your choice: "))
        
        if choice == 1:
            item = input("Enter product name: ")
            products.append(item)
            print(f"{item} added successfully.")
        
        elif choice == 2:
            item = input("Enter product name to remove: ")
            if item in products:
                products.remove(item)
                print(f"{item} removed successfully.")
            else:
                print("Product not found.")
        
        elif choice == 3:
            item = input("Enter product name to search: ")
            if item in products:
                print(f"{item} is available.")
            else:
                print("Product not found.")
        
        elif choice == 4:
            print("Product List:", products)
        
        elif choice == 5:
            print("Exiting program...")
            break
        
        else:
            print("Invalid choice. Please select from 1–5.")
    
    except ValueError:
        print("Error: Please enter a valid number.")
```

```python
while True:
    try:
        num = int(input("Enter a number: "))
        if num < 0:
            print("Negative number entered. Program stopped.")
            break
        if num % 4 == 0:
            continue
        print("You entered:", num)
    
    except ValueError:
        print("Invalid input. Please enter a valid integer.")
```

```python
try:

    with open("numbers.txt", "w") as f:
        f.write("12\n45\n7\n89\n23\n")

    with open("numbers.txt", "r") as f:
        numbers = [int(line.strip()) for line in f]

    total = sum(numbers)

    largest = max(numbers)
    smallest = min(numbers)

    print("Numbers:", numbers)
    print("Sum:", total)
    print("Largest:", largest)
    print("Smallest:", smallest)

except FileNotFoundError:
    print("Error: numbers.txt file not found!")
```
